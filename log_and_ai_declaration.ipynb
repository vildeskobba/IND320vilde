# Log 

For this assignment in IND320, I worked on developing a Streamlit app consisting of four pages. The main goal was to create an interactive way of presenting and exploring data. I made a homepage that contains buttons linking to the different pages, as well as a page displaying a dataframe and another page containing different plots. I enjoyed experimenting with the visual side of the application, especially the styling and colors in Streamlit. It was interesting to see how small changes in the design could make the app feel more personal and organized. I organized the different pages into a separate folder, which made the project easier to navigate.

In addition to the Streamlit app, I worked on a Jupyter Notebook where I scaled and plotted the data. This helped me understand the data better before presenting it in the application. I also focused on keeping the project structured and reusable. 

I created another folder called `styling`, where I placed my styling functions and colors. The colors are organized in a dictionary, making it easier to access and reuse them throughout the application. This means that if I want to change the overall color scheme later, I can do so in one place instead of changing individual colors throughout the code.

I also used a `utils.py` file for functions that are used on multiple pages. For example, I created a home button that appears on every page. Instead of writing the same code repeatedly, I put the function in `utils.py`. I organized the button functionality in a class called `Buttons`, which also makes it easier to add more buttons in the future.

Overall, I enjoyed working on this assignment because it combined programming, data analysis, and design. I especially liked experimenting with the styling and decided to make the buttons pink because that makes me happy.
